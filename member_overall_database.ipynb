### Loan Informations

```python
# loan information
query ="""
WITH filtered_members AS (
SELECT zones.name zonesName, zones.pmuk_id pmuk_ids, members.id as member_id, admission_date from members
LEFT JOIN shamities on shamities.id = members.shamity_id
JOIN zones
    ON zones.pmuk_id = LEFT(shamities.pmuk_id,2)
-- where shamities.pmuk_id like '01%' and admission_date BETWEEN '2026-02-01' AND '2026-04-30'
),

filtered_schedules AS (
    SELECT
    zonesName,
    pmuk_ids,
    filtered_members.member_id as mem_id,admission_date,
        ls.*
    FROM filtered_members
    LEFT JOIN loan_schedules ls on ls.member_id = filtered_members.member_id
    -- WHERE ls.due_date BETWEEN '2026-02-01' AND '2026-04-30'
)

SELECT
    zonesName,
    pmuk_ids,
    ls.mem_id,
    tl.shamity_id,

    lc.name AS loan_component,

    l.loan_product_id,
    l.id AS loan_id,
    l.disbursed_date,
    l.loan_amount,
    l.total_interest,
    l.balance,

    ls.id AS schedule_id,
    tl.id AS transaction_id,

    l.next_unpaid_schedule_id,
    l.is_active AS loan_activity,
    l.loan_condition,
    l.status AS loan_status,

    tl.transaction_type,
    l.overdue,

    -- schedule
    ls.kisti_no,
    ls.due_date,
    tl.date_of_action AS transaction_date,
    ls.kisti_amount,
    ls.principal_amount,
    ls.interest_amount,
    ls.paid_amount,
    tl.amount AS transaction_amount,
    ls.paid_principal,
    ls.paid_interest,
    ls.kisti_paid_date,
    ls.kisti_status,
    ls.is_holiday AS holiday,

    -- transaction loans
    tl.pre_sthiti,
    tl.pre_bokeya_sthiti,
    tl.post_sthiti,
    tl.post_bokeya_sthiti,
    tl.principal,
    tl.service_charge

FROM filtered_schedules ls

 INNER JOIN loans l
    ON l.id = ls.loan_id

LEFT JOIN transactions_loans tl
    ON tl.loan_id = l.id
    AND tl.loan_schedule_id = ls.id

LEFT JOIN loan_products lp
    ON lp.id = l.loan_product_id

LEFT JOIN loan_components lc
    ON lc.id = lp.component_id
where  l.id = 2123

"""
```

### Savings_Informations

```python
# savings
savings ="""
WITH aggregated AS (
    SELECT
        t.member_id,
        SUM(CASE WHEN t.savings_type = 'rspw' AND t.transaction_type <> 'withdraw' THEN t.amount ELSE 0 END) AS rspw_deposit,
        SUM(CASE WHEN t.savings_type = 'rspw' AND t.transaction_type = 'withdraw' THEN t.amount ELSE 0 END) AS rspw_withdraw,
        SUM(CASE WHEN t.savings_type = 'rspm' AND t.transaction_type <> 'withdraw' THEN t.amount ELSE 0 END) AS rspm_deposit,
        SUM(CASE WHEN t.savings_type = 'rspm' AND t.transaction_type = 'withdraw' THEN t.amount ELSE 0 END) AS rspm_withdraw,
        SUM(CASE WHEN t.savings_type = 'vsp' AND t.transaction_type <> 'withdraw' THEN t.amount ELSE 0 END) AS vsps_deposit,
        SUM(CASE WHEN t.savings_type = 'vsp' AND t.transaction_type = 'withdraw' THEN t.amount ELSE 0 END) AS vsps_withdraw
    FROM transactions_savings t
    JOIN shamities s ON s.id = t.shamity_id
    WHERE s.pmuk_id LIKE '0102%'
      AND t.date_of_action BETWEEN '2026-02-01' AND '2026-04-30'
    GROUP BY t.member_id
),

latest_values AS (
    SELECT
        t.member_id,
        MAX(CASE WHEN t.savings_type = 'rspw' THEN t.post_sthiti ELSE 0 END) AS post_sthiti_rspw,
        MAX(CASE WHEN t.savings_type = 'rspm' THEN t.post_sthiti  ELSE 0 END) AS post_sthiti_rspm,
        MAX(CASE WHEN t.savings_type = 'vsp' THEN t.post_sthiti  ELSE 0 END) AS post_sthiti_vsp
    FROM (
        SELECT
            t.member_id,
            t.savings_type,
            t.post_sthiti,
            t.date_of_action,
            ROW_NUMBER() OVER (PARTITION BY t.member_id, t.savings_type ORDER BY t.date_of_action DESC) AS rn
        FROM transactions_savings t
        JOIN shamities s ON s.id = t.shamity_id
        WHERE s.pmuk_id LIKE '0102%'
    ) t
    WHERE rn = 1
    GROUP BY t.member_id
)

SELECT
    a.member_id,
    a.rspm_deposit,
    a.rspm_withdraw,
    l.post_sthiti_rspm,
    a.rspw_deposit,
    a.rspw_withdraw,
    l.post_sthiti_rspw,
    a.vsps_deposit,
    a.vsps_withdraw,
    l.post_sthiti_vsp
FROM aggregated a
LEFT JOIN latest_values l ON a.member_id = l.member_id
"""
```

```python
savings_info=pl.read_database(savings, conn)
savings_info
```
